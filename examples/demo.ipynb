Run the cells from top to bottom. A C11 `clang` must be on PATH.

## 1. A small C program

```python
%%writefile hello.c
#include <stdio.h>

static int square(int n) { return n * n; }

int main(void) {
    for (int i = 1; i <= 5; i++) {
        printf("%d 的平方是 %d\n", i, square(i));
    }
    return 0;
}
```

## 2. Arrange it as a cat from the command line

```python
!codeart --program hello.c --shape cat --output hello_cat.c
!cat hello_cat.c
```

## 3. It still compiles and prints the same output

```python
!clang -std=c11 hello.c -o hello && ./hello > before.txt
!clang -std=c11 hello_cat.c -o hello_cat && ./hello_cat > after.txt
!diff before.txt after.txt && echo "Same output" && cat after.txt
```

## 4. The Python API: all three shapes

```python
from codeart.contracts import GenerationRequest, ShapeSpec
from codeart.pipeline import generate

source = open("hello.c", encoding="utf-8").read()
for shape in ("circle", "cat", "code"):
    spec = ShapeSpec(shape, "CODE" if shape == "code" else None)
    result = generate(GenerationRequest(source, spec, program_mode=True))
    checks = {name: getattr(result.verification, name + "_status").value
              for name in ("token", "geometry", "compile")}
    print(shape, result.status.value, checks)
    print(result.rendered.artwork_source_text)
```
